### Question 1: Write a function called greet that takes a name as a parameter and returns "Hello, [name]!"

```python
def greet(name):
    return(f'Hello, {name}')
```

```python
greet ("Clayton")
```

### Question 2: Create a function calculate_area that takes length and width as parameters and returns the area of a rectangle.

```python
def calculate_area (length, width):
    area_of_triangle = 1/2*length*width
    return area_of_triangle
```

```python
calculate_area (10,15)
```

### Question 3: Define a function is_even that takes a number and returns True if it's even, False if it's odd.

```python
def is_even (number):
    if number%2==0:
        return ("true")
    else:
        return("false")
```

```python
is_even (3)
```

### Question 4: Write a function get_max that takes two numbers and returns the larger one.

```python
def get_max(a,b):
    if a>b:
        return a
    else:
        return b
```

```python
get_max(5,10)
```

### Question 5: Create a function celsius_to_fahrenheit that converts Celsius to Fahrenheit using the formula: F = (C × 9/5) + 32

```python
def celsius_to_fahrenheit (c):
    f=(c * 9/5)+32
    return f
```

```python
celsius_to_fahrenheit (20)
```

### Question 6: Write a function introduce that takes a name and an optional age parameter (default value 25). It should return "Hi, I'm [name] and I'm [age] years old."

### Question 7: Create a function calculate_discount that takes a price and discount percentage (default 10%) and returns the discounted price.

```python
def calculate_discount (price, discount_percentage=0.1):
    discount_price = price*(discount_percentage*1/100)
    return discount_price
```

```python
calculate_discount (300, 50)
```

### Question 8: Define a function create_email that takes a first name, last name, and optional domain (default "gmail.com") to create an email address.

```python
def create_email (first_name, last_name, domain="gmail.com"):
    email_address = first_name+last_name
    return (f'{email_address}@{domain}')
```

```python
create_email("datascientist", "clay")
```

### Question 9: Create a function count_vowels that takes a string and returns the number of vowels in it.

```python
def count_vowels (write_up):
    number_of_vowels = 0
    for i in write_up:
        if i.lower() in "aeiou":
            number_of_vowels += 1
    return number_of_vowels
```

```python
count_vowels ("freshman")
```

### Question 10: Define a function find_longest_word that takes a list of words and returns the longest one.

```python
def find_longest_word(sentence):
    words = sentence.split()
    longest_word = max(words, key=len)
    return longest_word
    
```

```python
find_longest_word("The teacher is going to be early today")
```
