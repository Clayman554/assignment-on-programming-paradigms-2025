### Question 3: Given a list of numbers, find and print the largest and smallest values

```python
list_of_numbers = [2,10,50,40,22,65,9,63,1,99]
largest=list_of_numbers[0]
smallest=list_of_numbers[0]
for number in list_of_numbers:
    if number>largest:
        largest=number
for number in list_of_numbers:
    if number<smallest:
        smallest=number
print (f"The largest number is {largest} and the smallest number is {smallest}")

```

### Question 4: Write a program that removes all duplicate values from a list and prints the unique elements.

```python
list_of_numbers = [1,2,3,1,4,5,6,7,5,8]
no_duplicate=[]
for numbers in list_of_numbers:
    if numbers not in no_duplicate:
        no_duplicate.append(numbers)
print(no_duplicate)
    
```

### Question 5: Write a program that finds the most common element in a list and how many times it appears

```python
numbers = [8, 8, 2]
counts = {}
for number in numbers:
    if number in counts:
        counts[number] += 1
    else:
        counts[number] = 1
common_number = 0
max_count = 0

for number, count in counts.items():
    if count > max_count:
        max_count = count
        common_number = number

print(f"The common number is {common_number} and it appears {max_count} times.")
```

### Question 6: Create a program that counts the number of vowels and consonants in a sentence

```python
sentence = ("I am that good")
vowels = "aeiou"
vowel_count = 0
consonant_count = 0
for letter in sentence:
    if letter.isalpha():
        if letter in vowels:
            vowel_count += 1
        else:
            consonant_count += 1
print(f'The vowel count is {vowel_count}')
print(f'The consonant count is {consonant_count}')
```

### Question 7: Write a program that finds the longest word in a sentence entered by the user.

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
longest_word = ("")
for word in words:
    if len(word) > len(longest_word):
        longest_word = word
print(f"The longest word is: {longest_word}")
```

### Question 9: Create a program that sums user-entered numbers until the total exceeds 100, then stops and reports how many numbers were entered.

```python
total_number = 0
count = 0
while total_number <= 100:
    number = int(input("Enter a number: ")) 
    total_number += number                        
    count += 1                               
print("The total is now over 100!")
print("You entered", count, "numbers.")
```

### Question 10: Write a program using a while loop that keeps asking for a password until the user enters "secret123".

```python
correct_password = ("secret123")
customers = input("write a password: ")
while customers != correct_password:
    customers = input("write a password: ")
else:
    print("Access granted!")
```

### Question 13: Write a program that prints the multiplication table of a number entered by the user up to 10.

### Question 14: Write a program that asks the user to enter numbers repeatedly and stops when the user enters a negative number, then prints the list of entered numbers

```python
numbers = []
while True:
    num = int(input("insert a number: "))
    if num < 0:
        break
    numbers.append(num)
print(f"You typed these numbers:{numbers}")
```

### Question 15: Write a program that keeps asking for a number until the user enters 0, then outputs the sum of all entered numbers.

```python
total = 0
while True:
    num = float(input("Type a number: "))
    if num == 0:
        break
    new_total = total + num 
    total = new_total
print("Total is:", total)
```
